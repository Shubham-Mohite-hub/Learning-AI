```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
df = pd.read_csv('IPL.csv')
df.head()
```

```python
df.info()
```

```python
print(f"The size of the row is {df.shape[0]} and size of coloumn is {df.shape[1]}")
```

```python
df.isnull().sum()
```

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index, x = match_wins.values, palette='viridis')
plt.title("Most match win by the team")
```

Toss Decision Trends

```python
sns.countplot(x = df['toss_decision'], palette='rainbow')
plt.title("Toss Decision Trends")
```

Toss Winner vs match Winner

```python

count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = (count *100)/df.shape[0]
percentage.round(2)
```

How do teams win? (Runs vs Wickets)

```python
sns.countplot(x = df['won_by'])
plt.title("Won By")
```

Most "Player of the Match" Awards

```python
count1 = df['player_of_the_match'].value_counts().head(10)
count1
```

```python
sns.barplot(x = count1.values, y = count1.index, palette='mako')
plt.title("Top 10 players with man of the match")
```

2 Top Scorers

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending =False).head(2)
high
```

```python
high.plot(kind =  'barh')
```

10 Best Bowling Figures

```python
df.head()
```

```python
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x:x.split('--')[0])
df['highest_wickets'] = df['highest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending = False).head(10)
top_bowlers.plot(kind = 'barh')
```
