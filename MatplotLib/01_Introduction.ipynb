```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

MatplotLib Basics

```python
x = np.linspace(0,5,11)
x
y = x**3
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

Subplots in matplotLib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x**x, x**x)
plt.subplot(2,2,4)
plt.plot(x, x**x)
```

Object Oriented Way

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axis1.plot(x, y)
axis2 = fig.add_axes([0.3, 0.3, 0.4, 0.4])
axis2.plot(x, y)
```

```python
plt.scatter(x,y)
```
